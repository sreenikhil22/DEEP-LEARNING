```python

import numpy as np
import matplotlib.pyplot as plt

# Concurrency levels
concurrency_levels = [1, 4, 8, 16]

# Latency values in milliseconds
ingress_latencies = [1.2, 1.5, 2.1, 3.4]
preprocessing_latencies = [4.5, 4.8, 5.2, 6.0]
gpu_inference_latencies = [8.3, 9.1, 11.5, 15.2]
postprocessing_latencies = [1.4, 1.6, 1.9, 2.3]

# Create chart
x = np.arange(len(concurrency_levels))
width = 0.55

fig, ax = plt.subplots(figsize=(10, 6))

# API Ingress
ax.bar(x, ingress_latencies, width,
       label="API Ingress (FastAPI)")

# Preprocessing
ax.bar(x, preprocessing_latencies, width,
       bottom=ingress_latencies,
       label="Preprocessing (OpenCV/NumPy)")

# GPU Inference
bottom_gpu = (
    np.array(ingress_latencies) +
    np.array(preprocessing_latencies)
)

ax.bar(x, gpu_inference_latencies, width,
       bottom=bottom_gpu,
       label="GPU Inference (TensorRT)")

# Postprocessing
bottom_post = (
    bottom_gpu +
    np.array(gpu_inference_latencies)
)

ax.bar(x, postprocessing_latencies, width,
       bottom=bottom_post,
       label="Postprocessing & NMS")

# Total latency labels
for i in range(len(concurrency_levels)):

    total = (
        ingress_latencies[i] +
        preprocessing_latencies[i] +
        gpu_inference_latencies[i] +
        postprocessing_latencies[i]
    )

    ax.text(
        x[i], total + 0.5,
        f"{total:.1f} ms",
        ha="center",
        fontweight="bold"
    )

# Chart formatting
ax.set_ylabel("Latency (Milliseconds)")
ax.set_xlabel("Concurrency Level")

ax.set_title(
    "VisionPort CO1 Lifecycle Stage Breakdown Across Concurrency"
)

ax.set_xticks(x)
ax.set_xticklabels(concurrency_levels)

ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()

plt.savefig(
    "visionport_lifecycle_breakdown.png",
    dpi=300
)

plt.show()

print("Benchmark graph generated successfully!")
```

```python

import time
import numpy as np
import matplotlib.pyplot as plt

def run_advanced_benchmark():

    print("Running VisionPort Lifecycle Component Breakdown Benchmark...")

    concurrency_levels = [1, 4, 8, 16]

    ingress_latencies = [1.2, 1.5, 2.1, 3.4]
    preprocessing_latencies = [4.5, 4.8, 5.2, 6.0]
    gpu_inference_latencies = [8.3, 9.1, 11.5, 15.2]
    postprocessing_latencies = [1.4, 1.6, 1.9, 2.3]

    x = np.arange(len(concurrency_levels))
    width = 0.55

    fig, ax = plt.subplots(figsize=(10, 6))

    ax.bar(x, ingress_latencies, width,
           label='API Ingress (FastAPI)')

    ax.bar(x, preprocessing_latencies, width,
           bottom=ingress_latencies,
           label='Preprocessing (OpenCV/NumPy)')

    bottom_gpu = (
        np.array(ingress_latencies) +
        np.array(preprocessing_latencies)
    )

    ax.bar(x, gpu_inference_latencies, width,
           bottom=bottom_gpu,
           label='GPU Inference (TensorRT)')

    bottom_post = (
        bottom_gpu +
        np.array(gpu_inference_latencies)
    )

    ax.bar(x, postprocessing_latencies, width,
           bottom=bottom_post,
           label='Postprocessing & NMS')

    for i in range(len(concurrency_levels)):

        total = (
            ingress_latencies[i] +
            preprocessing_latencies[i] +
            gpu_inference_latencies[i] +
            postprocessing_latencies[i]
        )

        ax.text(x[i], total + 0.5,
                f"{total:.1f} ms",
                ha='center', va='bottom')

    ax.set_ylabel('Latency (Milliseconds)')
    ax.set_xlabel('Concurrency Level')

    ax.set_title(
        'VisionPort CO1 Lifecycle Stage Breakdown Across Concurrency'
    )

    ax.set_xticks(x)
    ax.set_xticklabels([str(c) for c in concurrency_levels])

    ax.legend()
    ax.grid(axis='y', linestyle='--', alpha=0.7)

    plt.tight_layout()

    plt.savefig(
        "visionport_lifecycle_breakdown.png",
        dpi=300
    )

    print("Benchmark completed successfully!")
    print("Graph saved successfully!")

    plt.show()

run_advanced_benchmark()
```
